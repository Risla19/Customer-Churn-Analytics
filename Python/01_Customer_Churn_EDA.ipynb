```python
import matplotlib.pyplot as plt

df["Churn"].value_counts().plot(kind="bar")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.show()
```

```python
churn_by_contract = df.groupby("Contract")["Churn"].value_counts(normalize=True).unstack() * 100

churn_by_contract
```

```python
churn_by_contract["Yes"].plot(kind="bar")

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
churn_by_internet = df.groupby("InternetService")["Churn"].value_counts(normalize=True).unstack() * 100

churn_by_internet
```

```python
churn_by_internet["Yes"].plot(kind="bar")

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
churn_by_payment = df.groupby("PaymentMethod")["Churn"].value_counts(normalize=True).unstack() * 100

churn_by_payment
```

```python
churn_by_payment["Yes"].plot(kind="bar")

plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
churn_by_support = df.groupby("TechSupport")["Churn"].value_counts(normalize=True).unstack() * 100

churn_by_support
```

```python
churn_by_support["Yes"].plot(kind="bar")

plt.title("Churn Rate by Tech Support")
plt.xlabel("Tech Support")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 Months", "13-24 Months", "25-48 Months", "49-72 Months"]
)

churn_by_tenure = df.groupby("TenureGroup", observed=False)["Churn"].value_counts(normalize=True).unstack() * 100

churn_by_tenure
```

```python
churn_by_tenure["Yes"].plot(kind="bar")

plt.title("Churn Rate by Customer Tenure")
plt.xlabel("Customer Tenure")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()
```
